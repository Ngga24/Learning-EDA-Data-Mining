# 1. Persiapan Environment & Import Library

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
from google.colab import drive
```

```python
drive.mount('/content/drive')
file_path = '/content/drive/MyDrive/Colab Notebooks/titanic_train_clean.csv'

df_clean = pd.read_csv(file_path)

# Menampilkan sempel 5 baris
df_clean.head()
```

```python
# Mengatur tampilan default agar grafik terlihat rapi
sns.set_theme(style="whitegrid")
plt.rcParams["font.family"] = "sans-serif"
```

# 2. Memeriksa Ringkasan Statistik Data Secara Menyeluruh

```python
# Cek informasi kolom dan tipe data
print(df_clean.info())

# Tampilkan ringkasan statistik numerik
display(df_clean.describe())
```

# 3. Visualisasi Distribusi Variable Numerik

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Distribusi Usia
sns.histplot(df_clean['Age'], kde=True, ax=axes[0], color='teal', bins=20)
axes[0].set_title('Distribusi Usia Penumpang (Hasil Imputasi)')
axes[0].set_xlabel('Usia')

# Perbandingan Fare Mentah vs Fare_Log
sns.kdeplot(df_clean['Fare_Log'], ax=axes[1], color='coral', fill=True)
axes[1].set_title('Distribusi Fare setelah Transformasi Log')
axes[1].set_xlabel('Fare_Log')
axes[1].set_xlim(left=-0.5)

plt.tight_layout()
plt.show()
```

# 4. Visualisasi Fitur Kategori & Hasil Engineering

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# Visualisasi Kolom Title Hasil One-Hot Encoding
# Mengambil semua kolom yang diawali nama 'Title_'
title_cols = [col for col in df_clean.columns if col.startswith('Title_')]

if title_cols:
    title_counts = df_clean[title_cols].sum().reset_index()
    title_counts.columns = ['Title', 'Count']
    # Merapikan label nama (menghapus prefix 'Title_')
    title_counts['Title'] = title_counts['Title'].str.replace('Title_', '')

    sns.barplot(data=title_counts, x='Title', y='Count', ax=axes[0], palette='Blues_r')
    axes[0].set_title('Frekuensi Gelar Penumpang (One-Hot Encoded)')
    axes[0].tick_params(axis='x', rotation=30)
else:
    # Fallback ke Pclass jika kolom Title_ tidak ditemukan
    sns.countplot(data=df_clean, x='Pclass', ax=axes[0], palette='Blues_r')
    axes[0].set_title('Distribusi Kelas Penumpang (Pclass)')

# Ukuran Keluarga (FamilySize)
sns.countplot(data=df_clean, x='FamilySize', ax=axes[1], palette='Greens_r')
axes[1].set_title('Distribusi Ukuran Keluarga')

# Proporsi Penumpang Sendirian (IsAlone)
sns.countplot(data=df_clean, x='IsAlone', ax=axes[2], palette='Oranges_r')
axes[2].set_title('Penumpang Sendirian (0 = Tidak, 1 = Ya)')

plt.tight_layout()
plt.show()
```

# 5. Visualisasi Matriks Korelasi (Heatmap)

```python
plt.figure(figsize=(16, 12))

# Buat Heatmap dengan font angka lebih kecil
sns.heatmap(
    df_clean.corr(),
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    linewidths=0.5,
    annot_kws={"size": 8}
)

plt.title("Matriks Korelasi Variabel Dataset Titanic", fontsize=16, pad=15)
plt.xticks(rotation=45, ha='right', fontsize=10)
plt.yticks(fontsize=10)
plt.tight_layout()
plt.show()
```
